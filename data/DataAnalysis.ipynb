# RouterBench data

Loads RouterBench (0-shot and 5-shot), looks at the data and writes two CSVs to read:
`df0_filtered.csv` (0-shot) and `df5_filtered.csv` (5-shot), with each model's score and cost.

The experiment scripts do not use these files; they use the output of `src/data/build_dataset.py`.
Do not re-save the CSVs from Excel: it changes the separator and garbles non-English text.

```python
try:
    import huggingface_hub  # noqa: F401
except ImportError:
    %pip install -q huggingface_hub
```

```python
from pathlib import Path

import pandas as pd
from huggingface_hub import hf_hub_download

FILES = {"0shot": "routerbench_0shot.pkl", "5shot": "routerbench_5shot.pkl"}

for name in FILES.values():
    if not Path(name).exists():
        hf_hub_download(
            repo_id="withmartian/routerbench", filename=name, repo_type="dataset", local_dir="."
        )
```

## 2. Models

RouterBench column keys of the models in this study. Delete a line to leave a model out.

```python
MODELS = {
    "gpt-4-1106-preview": "GPT-4",
    "gpt-3.5-turbo-1106": "GPT-3.5",
    "claude-v1": "Claude v1",
    "claude-v2": "Claude v2",
    "claude-instant-v1": "Claude Instant",
}
MODEL_KEYS = list(MODELS)
```

## 3. Problems the cleaning has to handle

```python
def problems(df):
    scores = df[MODEL_KEYS].apply(pd.to_numeric, errors="coerce")
    return pd.Series(
        {
            "rows": len(df),
            "unique sample_ids": df["sample_id"].nunique(),
            "rows with a missing score": int(scores.isna().any(axis=1).sum()),
            "extra rows of a repeated sample_id": int(df["sample_id"].duplicated().sum()),
            "extra rows of a repeated prompt": int(df["prompt"].astype(str).duplicated().sum()),
        }
    )


pd.DataFrame({"0-shot": problems(df0), "5-shot": problems(df5)})
```

```python
# Rows that share a sample_id, with each copy's prompt length.
repeated = df5[df5["sample_id"].duplicated(keep=False)].sort_values("sample_id")
repeated = repeated.assign(prompt_chars=repeated["prompt"].astype(str).str.len())
repeated[["sample_id", "prompt_chars"] + MODEL_KEYS].head(10)
```

In the 5-shot file, 14 ARC questions appear twice. GPT-3.5 was scored on a prompt about twice
the normal length, the other four models on the normal prompt. No single prompt has all five
scores, so the cleaning below drops both rows. The 0-shot file has these questions complete.

## 4. Clean and export

Keep each model's score and cost, drop rows with a missing value, keep one row per question.

```python
COLUMNS = ["sample_id", "prompt", "eval_name"] + MODEL_KEYS + [f"{key}|total_cost" for key in MODEL_KEYS]


def clean(df):
    out = df[COLUMNS].copy()
    out[MODEL_KEYS] = out[MODEL_KEYS].apply(pd.to_numeric, errors="coerce")
    out = out.dropna()  # rows with a missing score or cost
    out = out.drop_duplicates(subset="sample_id")  # one row per question
    return out.reset_index(drop=True)


df0_filtered = clean(df0)
df5_filtered = clean(df5)
print(f"0-shot: {len(df0)} -> {len(df0_filtered)} rows")
print(f"5-shot: {len(df5)} -> {len(df5_filtered)} rows")

df0_filtered.to_csv("df0_filtered.csv", index=False)
df5_filtered.to_csv("df5_filtered.csv", index=False)
```
